2. Melakukan import library yang dibutuhkan untuk keseluruhan proyek.

```python
# Import library
from google_play_scraper import reviews_all, reviews, Sort
import pandas as pd
```

3. Melakukan scraping untuk ulasan dari aplikasi Shoppe.

```python
# Mengambil 30.000 ulasan aplikasi Shopee
scrapreview = reviews(
    'com.shopee.id',  # Ganti dengan package name Shopee
    lang='id',         # Bahasa Indonesia
    country='id',      # Negara Indonesia
    sort=Sort.MOST_RELEVANT,  # Ulasan paling relevan
    count=30000,       # Jumlah ulasan yang ingin diambil
    filter_score_with=5  # Hanya ambil ulasan dengan rating 5 bintang
)
```

4. Menyimpan hasil menjadi file csv.

```python
# Menyimpan hasil scrap ke dalam file csv
data = pd.DataFrame(scrapreview[0])
data.to_csv('HasilScraping_ulasan_aplikasi_ Shopee.csv', index=False)
```
